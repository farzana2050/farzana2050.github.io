```python
import os
from CRUD_Python_Module import AnimalShelter

# Set the database password for this Codio session.
# Do not put this password in the CRUD_Python_Module.py source file.
os.environ["AAC_DB_PASSWORD"] = "YOUR_DATABASE_PASSWORD"

shelter = AnimalShelter()

# CREATE
test_data = {
    "animal_id": "TESTCRUD001",
    "name": "Project One Dog",
    "animal_type": "Dog",
    "breed": "Labrador Retriever Mix",
    "outcome_type": "Transfer"
}

create_result = shelter.create(test_data)
print("Create Result:", create_result)

# READ
read_result = shelter.read({"animal_id": "TESTCRUD001"})

print("\nRead Result:")
for record in read_result:
    print(record)

# UPDATE
update_result = shelter.update(
    {"animal_id": "TESTCRUD001"},
    {"name": "Updated Project One Dog"}
)

print("\nUpdate Result:", update_result)

updated_record = shelter.read(
    {"animal_id": "TESTCRUD001"}
)

print("\nUpdated Record:")
for record in updated_record:
    print(record)

# DELETE
delete_result = shelter.delete(
    {"animal_id": "TESTCRUD001"}
)

print("\nDelete Result:", delete_result)

deleted_record = shelter.read(
    {"animal_id": "TESTCRUD001"}
)

print("\nAfter Delete:")
print(deleted_record)

# Close the database connection
shelter.close()
```
